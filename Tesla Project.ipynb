The MONTH column is stored as object, so the values look like text instead of datetime objects



```python
df_tesla['MONTH'] = pd.to_datetime(df_tesla['MONTH'])
df_tesla.dtypes
```

converted the MONTH column from text to a proper datetime format using the pd.to_datetime function. This lets me work with the dates accurately and sort or filter the data without errors.

```python
df_tesla['year'] = df_tesla['MONTH'].dt.year

yearly_search = df_tesla.groupby('year')['TSLA_WEB_SEARCH'].sum()

print(yearly_search)
```

```python
import plotly.graph_objects as go

fig = go.Figure(data=[
    go.Bar(
        x=yearly_search.index, 
        y=yearly_search.values,
        marker_color='orange',  
        text=yearly_search.values,  
        textposition='auto'
    )
])

fig.update_layout(
    title='Yearly Tesla Search Volume',
    xaxis_title='Year',
    yaxis_title='Total Search Volume',
    title_font_size=20,
    xaxis_tickangle=-45,
    template='plotly_white', 
    width=900,
    height=500
)

fig.show()
```

This bar chart shows the total Tesla web searches per year. By aggregating monthly data into yearly totals, we can easily see how public interest in Tesla has changed over time. The chart highlights years with high search activity, giving a clear view of trends and spikes in popularity.

```python
plt.figure(figsize=(14,8), dpi=120) 
plt.title('Tesla Web Search vs Price', fontsize=18)
 
ax1 = plt.gca()
ax2 = ax1.twinx()
 
ax1.set_ylabel('TSLA Stock Price', color='#E6232E', fontsize=14)
ax2.set_ylabel('Search Trend', color='skyblue', fontsize=14)
 
ax1.plot(df_tesla.MONTH, df_tesla.TSLA_USD_CLOSE, color='#E6232E', linewidth=3)
ax2.plot(df_tesla.MONTH, df_tesla.TSLA_WEB_SEARCH, color='skyblue', linewidth=3)
```

This chart compares Tesla’s stock price with Google search interest over time. One line shows how the price moved. The other shows how people searched for Tesla. You see both trends rising together. It hints that higher public interest lines up with higher stock activity.
